# Q22

```python
top_directors = data.Director.value_counts().head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top_directors.values, y=top_directors.index, color="steelblue")

plt.title("Top 10 Directors by Number of Movies", fontsize=16, fontweight="bold")
plt.xlabel("Total Movies", fontsize=12)
plt.ylabel("Director", fontsize=12)

plt.tight_layout()
plt.show()
```

# Q36

```python
import matplotlib.pyplot as plt

drama = data[
    data.Genre.str.contains('Drama', na = False)
]['Gross'].dropna()

action = data[
    data.Genre.str.contains('Action', na = False)
]['Gross'].dropna()

plt.figure(figsize=(8, 5))
plt.boxplot([drama, action], tick_labels=['Drama', 'Action'])

plt.title('Gross Revenue: Drama vs Action')
plt.xlabel('Genre')
plt.ylabel('Gross Revenue')
plt.tight_layout()
plt.show()
```

# Q40

```python
top_movies = data.nlargest(10, 'IMDB_Rating').copy()

top_movies['IMDB_Rating_Scaled'] = top_movies['IMDB_Rating'] * 10

top_movies.plot(x='Series_Title',y=['Meta_score' , 'IMDB_Rating_Scaled'] , kind='bar' , figsize=(12, 6))

plt.title('Meta Score vs IMDb Rating (Scaled to 100)')
plt.xlabel('Movie')
plt.ylabel('Score out of 100')
plt.xticks(rotation=45, ha='right')
plt.legend(['Meta Score', 'IMDb Rating × 10'])
plt.tight_layout()
plt.show()
```
